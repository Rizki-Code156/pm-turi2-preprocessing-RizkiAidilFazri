#C. Langkah Praktikum


##Membuat Dataset Karyawan


```python
import pandas as pd
import numpy as np

data = {
    'usia': [23, 30, np.nan, 35, 28, 42, 31, np.nan, 26, 38],
    'pendapatan': [4000000, 5500000, 4500000, np.nan, 5000000,
                   7500000, 6000000, 4800000, np.nan, 7000000],
    'pendidikan': ['SMA', 'S1', 'S1', 'S2', 'SMA',
                   'S2', 'S1', 'SMA', 'S1', 'S2'],
    'kota': ['Cirebon', 'Bandung', 'Jakarta', 'Cirebon', 'Bandung',
             'Jakarta', 'Cirebon', 'Bandung', 'Jakarta', 'Cirebon'],
    'membeli': ['Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak',
                'Ya', 'Ya', 'Tidak', 'Tidak', 'Ya']
}

df = pd.DataFrame(data)

df
```

##Langkah 1 — Menangani nilai hilang (imputasi median/modus).

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

##Langkah 2 — Memisahkan fitur (X) dan label (y).

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

##Langkah 3 — Encoding kategorikal (ordinal & one-hot).

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
```

##Langkah 4 — Membagi data (SEBELUM scaling).

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)
```

##Langkah 5 — Penskalaan (fit pada latih, transform pada latih & uji).

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage
```

##Langkah 6 — Verifikasi: tidak ada nilai hilang, semua numerik, skala seragam.

```python
print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))
print("\nNilai yang hilang:")
print(X.isna().sum())
print("\nData X_train:")
display(X_train)
```

```python
from sklearn.preprocessing import MinMaxScaler

sc_minmax = MinMaxScaler()

X_train_minmax = X_train.copy()
X_test_minmax = X_test.copy()

X_train_minmax[num] = sc_minmax.fit_transform(X_train_minmax[num])
X_test_minmax[num] = sc_minmax.transform(X_test_minmax[num])

print("Hasil MinMaxScaler:")
display(X_train_minmax[num].describe())
```

##3. Tambahkan kolom kategorikal baru dan terapkan one-hot encoding.

```python
df_status = df.copy()

# Menambahkan kolom kategorikal baru
df_status['status'] = [
    'Belum Menikah',
    'Menikah',
    'Belum Menikah',
    'Menikah',
    'Belum Menikah',
    'Menikah',
    'Menikah',
    'Belum Menikah',
    'Belum Menikah',
    'Menikah'
]

# Melakukan One-Hot Encoding
df_status = pd.get_dummies(
    df_status,
    columns=['status'],
    dtype=int
)

# Menampilkan hasil
print("Data setelah menambahkan kolom status dan One-Hot Encoding:")
display(df_status)
```

Dengan demikian, data kategorikal dapat digunakan dalam proses pembelajaran mesin.

##4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

fit_transform digunakan pada data latih karena scaler perlu mempelajari parameter dari data latih terlebih dahulu, kemudian melakukan transformasi terhadap data tersebut.

Sedangkan pada data uji hanya digunakan transform karena data uji harus menggunakan parameter yang sudah diperoleh dari data latih. Hal ini dilakukan untuk mencegah data leakage, yaitu informasi dari data uji ikut memengaruhi proses preprocessing.

#E. Refleksi

##1. Mengapa urutan 'split dulu, baru scaling' penting?

Split data dilakukan terlebih dahulu sebelum scaling untuk mencegah data leakage. Jika scaling dilakukan sebelum data dibagi menjadi data training dan testing, informasi dari data testing dapat ikut digunakan dalam proses scaling. Dengan melakukan split terlebih dahulu, parameter scaling dipelajari hanya dari data training, kemudian diterapkan pada data testing.

##2. Kapan label encoding dan kapan one-hot encoding?

Label encoding digunakan untuk data kategorikal yang memiliki urutan atau tingkatan, seperti pendidikan SMA, S1, dan S2. Sedangkan one-hot encoding digunakan untuk data kategorikal yang tidak memiliki urutan, seperti nama kota Cirebon, Bandung, dan Jakarta.

##3. Apa perbedaan normalisasi dan standardisasi?

Normalisasi mengubah nilai data ke dalam rentang tertentu, misalnya 0 sampai 1 seperti yang dilakukan oleh MinMaxScaler. Sedangkan standardisasi mengubah data berdasarkan nilai rata-rata dan standar deviasi sehingga data umumnya memiliki rata-rata 0 dan standar deviasi 1 seperti pada StandardScaler.
